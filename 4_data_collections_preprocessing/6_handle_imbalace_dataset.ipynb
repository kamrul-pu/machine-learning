```python
# Imbalanced dataset -> Dataset with an unequal class distribution
```

```python
import pandas as pd
import numpy as np
```

```python
# loading credit card csv data to pandas dataframe
df = pd.read_csv("credit_data.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.tail()
```

```python
# determine the distribution of the two classes
df["Class"].value_counts()
```

```python
# This is highly imbalanced dataset
# 0 -> Legit Transactions
# 1 -> Fraudulent Transactions
```

```python
# Separating the legit and fraudulent transactions
legit = df[df.Class == 0]
fraud = df[df.Class == 1]
```

```python
legit.shape
```

```python
fraud.shape
```

```python
# Under-sampling (Take some data to balance the dataset
```

```python
# Build a sample dataset containing similar distribution
# Number of fraudulent transaction -> 492
```

```python
legit_sample = legit.sample(n=492) # we can take 500 or almost similar sample
```

```python
legit_sample.shape
```

```python
# concatenate two DataFrames
```

```python
new_df = pd.concat([legit_sample, fraud], axis=0)
```
