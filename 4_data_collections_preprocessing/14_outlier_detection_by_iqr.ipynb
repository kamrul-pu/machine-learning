```python
import pandas as pd
df = pd.read_csv("heights_2.csv")
df.head()
```

```python
df.describe()
```

```python
Q1 = df.height.quantile(0.25) # same as 25% of describe output
Q3 = df.height.quantile(0.75) # same as 75%
Q1, Q3
```

```python
IQR = Q3 - Q1
IQR
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
lower_limit, upper_limit
```

```python
# find outliers
df[(df.height>upper_limit) | (df.height<lower_limit)]
```

```python
# remove outliers
df1 = df[(df.height<upper_limit) & (df.height>lower_limit)]
df.shape, df1.shape
```
