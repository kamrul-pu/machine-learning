```python
# importing the dependent modules
import pandas as pd
from sklearn import tree
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("titanic.csv")
df.head()
```

```python
# df["Sex"] = df["Sex"].map({"male": 1, "female": 2})
df.Sex.isnull().sum()
```

```python
# Taking the necessary columns and removeing p id name sibsp parch ticket cabon and embarked
inputs = df.drop(["PassengerId", "Name", "SibSp", "Parch", "Cabin","Ticket", "Embarked", "Survived"], axis="columns")
target = df.Survived
```

```python
inputs.Age.isnull().sum()
```

```python
inputs.Age = inputs.Age.fillna(inputs.Age.mean())
```

```python
inputs.Age.isnull().sum()
```

```python
inputs.Sex = inputs.Sex.map({"male": 1, "female": 2})
```

```python
inputs.Sex
```

```python
inputs.isnull().sum()
```

```python
# Splitting the value into trin and test
X_train, X_test, Y_train, Y_test = train_test_split(inputs, target, test_size=0.1, stratify=target, random_state=2)
```

```python
inputs.shape, X_train.shape, X_test.shape
```

```python
model = tree.DecisionTreeClassifier()
```

```python
model.fit(X_train, Y_train)
```

```python
model.score(X_train, Y_train)
```

```python
model.score(X_test, Y_test)
```
