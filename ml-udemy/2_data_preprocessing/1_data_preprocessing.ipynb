```python
# importing the dependent libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# importing the dataset
dataset = pd.read_csv("Data.csv")
X = dataset.iloc[:, :-1].values # row 0 - n(all rows) columns first to second last 0 - n-1
y = dataset.iloc[:, -1].values # all row last columns
```

```python
print(X)
```

```python
print(y)
```

```python
# some information about the dataset
dataset.info()
```

```python
# statistical measures of the dataset
dataset.describe()
```

```python
# Taking care of missing data
# for large dataset delete missing row, for small dataset fill the missing  value mean mode median
# we will fill missing  age & salary by average salary by sklearn imputer
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")
imputer.fit(X[:,1:3]) # fill missing value of age and salary by mean ( all row, only age col 1 and salary col 2)
X[:, 1:3] = imputer.transform(X[:, 1:3])
```

```python
print(X)
```

```python
# Encoding categorical data Country column
# One hot Encoding
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
print(X)
```

```python
# Encoding dependent varibale purched column using label encoder
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
print(y)
```

```python
# Split the dataset into train and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=1)
```

```python
print(X_train)
```

```python
print(X_test)
```

```python
print(y_train)
```

```python
print(y_test)
```

```python
# Feature scaling
# Normalization ->  works when data distribution is normal range 0-1
# Standarisation -> works on every situation range from -3 to +3 standard deviation
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
# apply standardisation only features not encoded columns (vategorical variable) cause they are already within the range
X_train[:, 3:] = sc.fit_transform(X_train[:,3:])
X_test[:, 3:] = sc.fit_transform(X_test[:,3:])
```
