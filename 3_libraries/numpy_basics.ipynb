```python
# time taken by a ist
python_list = [i for i in range(100000)]
start_time = process_time()
python_list = [i+5 for i in range(100000)]
end_time = process_time()
print(f"time taken = {end_time-start_time}")
```

```python
numpy_array = np.array([i for i in range(100000)])
start_time = process_time()
numpy_array+=5
end_time = process_time()
print(f"time taken = {end_time-start_time}")
```
