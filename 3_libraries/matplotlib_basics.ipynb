```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)
z = np.cos(x)
```

```python
print(x)
```

```python
print(y)
```

```python
print(z)
```

```python
# sine wave
plt.plot(x,y)
plt.show()
```

```python
# cosine wave
plt.plot(x,z)
plt.show()
```

```python
# adding title x-axis and y-axis labels
plt.plot(x,y)
plt.xlabel("angle")
plt.ylabel("sine value")
plt.title("sine wave")
plt.show()
```

```python
# adding title x-axis and y-axis labels
plt.plot(x,y)
plt.xlabel("angle")
plt.ylabel("cos value")
plt.title("cos wave")
plt.show()
```

```python
# parabola
x = np.linspace(-10,10,20)
y = x**2
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,"r*")
plt.show()
```

```python
# Multiple plot
x = np.linspace(-5,5,50)
plt.plot(x, np.sin(x), "g-")
plt.plot(x, np.cos(x), "r--")
plt.show()
```

```python
# Bar plot
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
languages = ["Arabic", "English", "Spanish", "Latin", "German"]
people = [100,50,150,40,70]
ax.bar(languages, people)
plt.xlabel("Languages")
plt.ylabel("number of people")
plt.show()
```

```python
# pie chart
```

```python
fig1 = plt.figure()
ax = fig1.add_axes([0,0,1,1])
languages = ["Arabic", "English", "Spanish", "Latin", "German"]
people = [100,50,150,40,70]
ax.pie(people, labels=languages, autopct="%1.1f%%")
plt.show()
```

```python
# Scatter Plot
x = np.linspace(0,10, 30)
y = np.sin(x)
z = np.cos(x)
fig2 = plt.figure()
ax = fig2.add_axes([0,0,1,1])
ax.scatter(x,y, color="g")
ax.scatter(x,z, color="b")
plt.show()
```
